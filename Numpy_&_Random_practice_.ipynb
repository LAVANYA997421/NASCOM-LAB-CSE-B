```python
import numpy as np

pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),
    (np.array([1, 2, 3]), np.array([3, 2, 1])),
    (np.array([2, 0, 0]), np.array([0, 5, 0])),
]

similarities = []

for i, (u, v) in enumerate(pairs, start=1):

    l1_norm = np.linalg.norm(u, ord=1)


    l2_norm = np.linalg.norm(u)


    cosine_similarity = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

    similarities.append(cosine_similarity)

    print(f"Pair {i}:")
    print("L1 norm =", l1_norm)
    print("L2 norm =", l2_norm)
    print("Cosine similarity =", cosine_similarity)
    print()


most_similar = np.argmax(similarities) + 1
print("Most similar pair = Pair", most_similar)
```

```python
import numpy as np

M = np.array([[4., 2.],
              [2., 3.]])
P = np.array([[0., -1.],
              [1.,  0.]])

M_inv = np.linalg.inv(M)
trace_M = np.trace(M)

print("Inverse of M:")
print(M_inv)

print("\nTrace of M:")
print(trace_M)

identity = M @ M_inv
print("\nM @ inv(M):")
print(identity)

print("\nIs M @ inv(M) equal to Identity Matrix?")
print(np.allclose(identity, np.eye(2)))
```

```python
import numpy as np


square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)


S = np.array([[1.5, 0.0],
              [0.0, 0.5]])


t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])


scaled = S @ square
rotated = R @ square


print("Original square:")
print(square)

print("\nScaled corners:")
print(scaled)

print("\nRotated corners:")
print(rotated)
```

```python
def close_loop(pts):

    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python

for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```
