Try-Except-Else-Finally Structure

```python
try:
    num = int(input("Enter a number to divide 100: "))
    result = 100 / num
except ZeroDivisionError:
    print("Cannot divide by zero!")
except ValueError:
    print("Invalid input! Not a number.")
else:
    print("Result is:", result)
finally:
    print("Execution complete. This block always runs.")
```

Catching Multiple Exceptions

```python
try:
    a = int("ten")  # Causes ValueError
    b = 10 / 0      # Causes ZeroDivisionError
except (ValueError, ZeroDivisionError) as e:
    print("An error occurred:", e)
```

# 1.1 Generate a list of 100 simulated sensor readings.
#    Each reading should be a tuple: (timestamp_str, device_id, device_type, location, value).
#    - timestamp_str: Format as 'YYYY-MM-DD HH:MM:SS'
#    - device_id: A unique ID for each reading (e.g., 'DEV001', 'DEV002', etc., up to 100 unique IDs).
#    - device_type: Randomly chosen from DEVICE_TYPES.
#    - location: Randomly chosen from LOCATIONS.
#    - value: Random float within READING_RANGES for its device_type (round to 2 decimal places).
# Store these in a list called 'raw_sensor_data'.


```python
import random
from datetime import datetime, timedelta

# Example definitions (remove if already provided)
DEVICE_TYPES = ["Temperature", "Humidity", "Pressure"]
LOCATIONS = ["Room_A", "Room_B", "Room_C"]

READING_RANGES = {
    "Temperature": (15.0, 40.0),
    "Humidity": (20.0, 90.0),
    "Pressure": (950.0, 1050.0)
}

# Generate 100 simulated sensor readings
raw_sensor_data = []

start_time = datetime.now()

for i in range(100):
    timestamp = start_time + timedelta(minutes=i)
    timestamp_str = timestamp.strftime("%Y-%m-%d %H:%M:%S")

    device_id = f"DEV{i+1:03d}"

    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    reading = (
        timestamp_str,
        device_id,
        device_type,
        location,
        value
    )

    raw_sensor_data.append(reading)

# Display first 5 readings
for reading in raw_sensor_data[:5]:
    print(reading)
```

# 1.4 Create a dictionary named 'device_stats' where keys are 'device_type' and values are
#     a tuple: (min_value, max_value) observed for that device type across all readings.
#     Initialize with sensible large/small values.





```python
device_stats = {}

# Initialize dictionary
for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf'))

# Update min and max values
for timestamp, device_id, device_type, location, value in raw_sensor_data:
    min_value, max_value = device_stats[device_type]

    if value < min_value:
        min_value = value
    if value > max_value:
        max_value = value

    device_stats[device_type] = (min_value, max_value)

# Display the results
print(device_stats)
```

2.1 Write a function `get_readings_by_location(readings_list, target_location)`
#     that takes the 'raw_sensor_data' list and a 'target_location' string.
#     It should return a list of all readings (as tuples) for that specific location

```python
def get_readings_by_location(readings_list, target_location):
    result = []

    for reading in readings_list:
        if reading[3] == target_location:
            result.append(reading)

    return result


# Example usage
location = "Room_A"   # Change to any location in your data
location_readings = get_readings_by_location(raw_sensor_data, location)

print(f"Readings from {location}:")
for reading in location_readings:
    print(reading)
```


# 2.2 Write a function `calculate_average_temperature(temp_data_list)`
#     that takes the 'temperature_readings' list (list of dictionaries).
#     It should calculate and return the average temperature. Return 0.0 if the list is empty.


```python
def calculate_average_temperature(temp_data_list):
    if len(temp_data_list) == 0:
        return 0.0

    total = 0

    for reading in temp_data_list:
        total += reading["temperature_celsius"]

    average = total / len(temp_data_list)
    return round(average, 2)


# Example usage
avg_temp = calculate_average_temperature(temperature_readings)

print("Average Temperature:", avg_temp)
```
